## ✅ Check 1: Python Version

We need Python 3.10 or higher for this course.

Run the cell below and look for the green ✅ checkmark!

````python
import sys

print("Checking Python version...\n")

version_info = sys.version_info

if version_info.major == 3 and version_info.minor >= 10:
    print(f"✅ Python {version_info.major}.{version_info.minor} detected - this should work!")
else:
    print(f"❌ Python {version_info.major}.{version_info.minor} detected")
    print("   This course requires Python 3.10 or higher.")
    print("   Did you activate the conda environment?")
    print("   Run: conda activate prompt-eng")
````

---

## 📦 Check 2: Required Packages

Let's verify all course packages are installed.

````python
print("Checking required packages...\n")

required_packages = {
    'openai': 'OpenAI API client',
    'dotenv': 'Environment variables (python-dotenv)',
    'tiktoken': 'Token counting',
    'requests': 'HTTP requests',
    'pydantic': 'JSON validation'
}

missing = []

for package in required_packages:
    try:
        if package == 'dotenv':
            import dotenv
        else:
            __import__(package)
    except ImportError:
        missing.append(package)

if not missing:
    print(f"✅ All {len(required_packages)} packages installed correctly!")
else:
    print(f"❌ Missing packages: {', '.join(missing)}")
    print("\n   To fix:")
    print("   1. conda activate prompt-eng")
    print("   2. conda env update -f ../environment.yml")
    print("   3. Restart JupyterLab")
````

---

## 🔧 Check 3: Jupyter Kernel

Verify you're using the correct conda environment in Jupyter.

````python
import sys

print("Checking Jupyter kernel...\n")

if "prompt-eng" in sys.executable.lower() or "prompt-eng" in sys.prefix.lower():
    print("✅ Correct conda environment active!")
else:
    print("⚠️ Warning: May not be using the prompt-eng environment")
    print("\n   To fix:")
    print("   1. In JupyterLab menu: Kernel → Change Kernel")
    print("   2. Select 'Python 3 (ipykernel)'")
    print("   3. Re-run this notebook")
````

## 🔧 Troubleshooting

**If any checks didn't pass, here's how to fix them:**

### Python version wrong
- Make sure you ran: `conda activate prompt-eng`
- Close and restart JupyterLab
- Check: Kernel → Change Kernel → Python 3 (ipykernel)

### Packages missing

**🍎 macOS & 🐧 Linux:**
```bash
cd ~/courses/prompt-eng
conda activate prompt-eng
conda env update -f environment.yml
```

**🪟 Windows:**
```powershell
cd "$env:USERPROFILE\courses\prompt-eng"
conda activate prompt-eng
conda env update -f environment.yml
```

Then restart JupyterLab and restart the kernel.

### Wrong kernel
- Kernel → Change Kernel → Python 3 (ipykernel)
- Make sure it says "(prompt-eng)" in your terminal
- Restart JupyterLab after activating environment

**Still having issues?**
- Copy any error message and paste it into ChatGPT or Claude — they're great at debugging setup issues
- Re-watch the setup lecture
- Post to the Q&A with your error message and output
