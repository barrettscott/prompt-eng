## 🔧 Step 1: Setup

````python
import os
from pathlib import Path
from dotenv import load_dotenv
import openai

load_dotenv(dotenv_path=Path("..") / ".env")
client = openai.OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
MODEL = "gpt-5-mini"

print("✅ Ready!")
````

## 💡 The Solution: Helper Functions

A **helper function** wraps repetitive code into a reusable package.

**DRY = "Don't Repeat Yourself"**

- Fewer bugs, easier maintenance


---

## 🔧 Building ask_openai() - Step by Step

Let's build our helper function. 

**Important Performance Note:** 
Notice we do **not** initialize `client = OpenAI(...)` inside this function. We use the global `client` we created in Step 1. This is much faster because we don't have to reconnect to OpenAI every time we ask a question.

````python
# We don't need to import openai here because we did it in Step 1!

def ask_openai(prompt, model=MODEL):
    """Ask OpenAI a question using the global client."""
    
    try:
        response = client.responses.create(
            model=model,
            input=prompt
        )
        return response.output_text.strip()
        
    except openai.AuthenticationError:
        return "Error: Invalid API key. Check your .env file."
    except openai.RateLimitError:
        return "Error: Rate limit exceeded. Wait and try again."
    except openai.APIConnectionError:
        return "Error: Network issue. Check your internet."
    except openai.BadRequestError:
        return "Error: Bad request. Check model name."
    except Exception as e:
        return f"API Error: {str(e)}"

print("✅ ask_openai() function created!")
````

### Key Features:
1. **Uses Global Client:** Efficient connection reuse.
2. **Simple Interface:** One function for all API calls.
3. **Returns Text:** Returns the response so you can save it to a variable.
4. **Robust Errors:** Catches the most common issues gracefully.

## 🎯 Using the Helper Function

Now watch how easy it is to use!

### Example 1: Simple Question

````python
print("❓ Question: What is Python?")
print("\n💬 Response:\n")

print(ask_openai("What is Python?"))

print()
````

## 🛡️ Error Handling Demo

Our helper function handles errors gracefully. Instead of crashing your entire program with a messy traceback, it catches the issue and returns a readable error message.

**Let's test this safely.** We will intentionally ask for a model that doesn't exist (e.g., `"gpt-fake-model"`). This forces the API to return an error, proving that our `try/except` block works!

````python
print("🛡️ Testing Error Handling:\n")

# 1. The Happy Path (This works)
print("--- Valid Request ---")
result = ask_openai("Hello!")
print(f"Result: {result[:50]}...") # Print first 50 chars

# 2. The Error Path (Safe Simulation)
# We intentionally ask for a model that doesn't exist.
# This proves our try/except block prevents a crash!
print("\n--- Invalid Request (Simulated Error) ---")
error_result = ask_openai("Hello!", model="gpt-fake-model-123")

print(f"Result: {error_result}")

print("\n✅ Success! The program didn't crash despite the API error.")
````
