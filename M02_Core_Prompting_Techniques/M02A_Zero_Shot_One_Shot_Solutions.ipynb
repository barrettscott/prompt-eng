## 🔧 Step 1: Setup

```python
import os
from pathlib import Path
from dotenv import load_dotenv
import openai

load_dotenv(dotenv_path=Path("..") / ".env")

client = openai.OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

MODEL = "gpt-5-mini"


def ask_openai(prompt, model=MODEL):
    """Ask OpenAI a question using the global client."""
    
    try:
        response = client.responses.create(
            model=model,
            input=prompt
        )
        return response.output_text.strip()
        
    except openai.AuthenticationError:
        return "Error: Invalid API key. Check your .env file."
    except openai.RateLimitError:
        return "Error: Rate limit exceeded. Wait and try again."
    except openai.APIConnectionError:
        return "Error: Network issue. Check your internet."
    except openai.BadRequestError:
        return "Error: Bad request. Check model name."
    except Exception as e:
        return f"API Error: {str(e)}"


print("✅ ask_openai() function created!")
```

## Exercise 2: Intent Detection

**Task:** Classify user intent into categories such as: question, request, complaint, or feedback.

**Approach:** Use one-shot prompting with a clear example showing the desired format.

```python
user_messages = [
    "How do I reset my password?",
    "I'd like to cancel my subscription",
    "The app keeps crashing on my phone",
    "Love the new features!"
]

print("=== Intent Classification ===\n")

for i, msg in enumerate(user_messages, 1):
    prompt = f"""Classify the user's intent.

Example:
Message: "When will my order arrive?"
question

Now classify this message:
Message: "{msg}"
"""
    
    print(f"{i}. Message: {msg}")
    print(f"   Intent: ", end="")
    print(ask_openai(prompt))
    print("=" * 60 + "\n")
```

### 💡 Solution Notes

**Key elements of this solution:**
- Clear example showing message → intent format
- Concise output (just the intent category)
- Works for multiple intent types
- Easy to parse programmatically

**Typical results:**
- "How do I reset my password?" → question
- "I'd like to cancel my subscription" → request
- "The app keeps crashing" → complaint
- "Love the new features!" → feedback

## Exercise 3: Code Comment Generator

**Task:** Generate concise inline comments for code snippets.

**Approach:** Use one-shot prompting to show the desired comment style.

```python
code_snippet = "result = [x**2 for x in range(10) if x % 2 == 0]"

prompt = f"""Generate a concise inline comment for code.

Example:
Code: data = sorted(items, key=lambda x: x['date'], reverse=True)
Sort items by date in descending order

Now generate a comment for this code:
Code: {code_snippet}
"""

print("=== Code Comment Generation ===\n")
print(f"Code: {code_snippet}")
print(f"\nComment: # ", end="")
print(ask_openai(prompt))
```

### 💡 Solution Notes

**Key elements of this solution:**
- Example shows code → comment format
- Comment is concise (one line)
- Explains what the code does, not how
- No "Comment:" prefix in output (we add the `#` ourselves)

**Expected result:**
Something like: "Create list of squares of even numbers from 0 to 9"
