## 🔧 Step 1: Setup

````python
import os
from pathlib import Path

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(dotenv_path=Path("..") / ".env")
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
MODEL = "gpt-5-mini"

print("✅ Ready!")
````

---

## 📚 Understanding the Responses API

The **Responses API** is OpenAI's modern interface for interacting with AI models. It simplifies the code by clearly separating your input from the model's instructions.

### Core Structure:
```python
response = client.responses.create(
    model=MODEL,                    # 1. Which AI model to use
    input="Your prompt here",       # 2. Your question, task, or prompt
    instructions="Be concise",      # 3. Optional: How the AI should respond
)
```

### Getting the Response:
```python
answer = response.output_text  # The AI's response as text
```

Let's see this in action!

---

## 🎯 Example 1: Simple Question with Error Handling

**Best Practice:** Always wrap API calls in a `try...except` block. This ensures your application handles network issues, authentication failures, or billing limits gracefully without crashing.

````python
simple_prompt = "What is Python programming?"

try:
    # Make the API call
    response = client.responses.create(
        model=MODEL,
        input=simple_prompt
    )
    
    # Get and display the answer
    print(f"Question: {simple_prompt}\n")
    answer = response.output_text
    print(f"✅ Answer: {answer}\n")
    
except Exception as e:
    print(f"❌ Error: {e}")
````

---

## 💰 Understanding Tokens & Costs

**Most students worry about API costs. Let's put that anxiety to rest!**

APIs charge based on **tokens** (roughly 3-4 characters per token, but it varies).

Let's see exactly what your API calls cost:

````python
prompt = "Explain the concept of tokenization in LLMs to a novice."

# Pricing per 1M tokens (check OpenAI pricing page for updates)
MODELS = {
    "gpt-4o-mini": {"input": 0.15, "output": 0.60},    
    "gpt-5-mini":  {"input": 0.25, "output": 2.00} 
}

try:
    response = client.responses.create(
        model=MODEL,
        input=prompt
    )
    
    # Get actual token counts from API (source of truth for billing)
    input_tokens = response.usage.input_tokens
    output_tokens = response.usage.output_tokens
    total_tokens = response.usage.total_tokens
    
    print("📊 Token Usage (Reported by API)")
    print("=" * 60)
    print(f"📝 Input Tokens:  {input_tokens}")
    print(f"🤖 Output Tokens: {output_tokens}")
    print(f"   Total Tokens:  {total_tokens}")
    print("=" * 60)
    
    # Show cost for both models
    print("\n💵 Cost Comparison (this request):")
    for model_name, pricing in MODELS.items():
        input_cost = (input_tokens / 1_000_000) * pricing["input"]
        output_cost = (output_tokens / 1_000_000) * pricing["output"]
        total_cost = input_cost + output_cost
        print(f"   {model_name}: ${total_cost:.6f}")
    
    print("\n💬 Response:")
    print(response.output_text)
    
except Exception as e:
    print(f"❌ Error: {e}")
````

### 💡 Cost Reality Check

<div style="text-align: left; display: inline-block;">

| What You're Doing         | gpt-4o-mini | gpt-5-mini |
|-------------------------|-------------|------------|
| Simple question         | \$0.0005    | \$0.002    |
| Summarize article       | \$0.001     | \$0.005    |
| Complete this notebook  | \$0.01      | \$0.05     |
| **Entire course** | ~\$1-2      | **~\$3-5** |

</div>

**Why show both?** We use `gpt-5-mini` throughout the course, but `gpt-4o-mini` works for all exercises if you want to minimize costs.

**Key Takeaway:** Either way, costs are minimal—experiment freely!

## 📝 Example 2: Using Instructions

The `instructions` parameter sets the AI's behavior, persona, and constraints.

````python
# Same input, different instructions
input_text = "What is Python programming?"

try:
    # Version 1: Professional and concise
    response1 = client.responses.create(
        model=MODEL,
        input=input_text,
        instructions="Answer in exactly one sentence and use a professional tone."
    )
    print("\n📌 Instructions: 'Answer in exactly one sentence...' ")
    print(f"Output: {response1.output_text}\n")
    
    # Version 2: Casual and simple
    response2 = client.responses.create(
        model=MODEL,
        input=input_text,
        instructions="Explain the concept to a 10-year-old using a fun analogy."
    )
    print("📌 Instructions: 'Explain to a 10-year-old...' ")
    print(f"Output: {response2.output_text}\n")
    
except Exception as e:
    print(f"❌ Error: {e}")
````

**Key Takeaway:** Use `instructions` for global rules (how to act), and the `input` for the specific task!

## 🔧 Troubleshooting

**API call fails with authentication error?**
- Check your `.env` file has the correct API key
- Verify the key is valid in your OpenAI dashboard
- Ensure `load_dotenv()` ran before the API call

**Response is empty or unexpected?**
- Check your `input` parameter has content
- Try simpler instructions first
- Print the full response object to debug

**Cost seems higher than expected?**
- Long outputs use more tokens than inputs
- Use `response.usage` to see actual token counts
- Check you're using `gpt-5-mini` not a larger model

**Still having issues?**
- Copy any error message and paste it into ChatGPT, Claude, Gemini, or Grok — they're great at debugging
- Re-watch the lecture for this module
- Post to the Q&A with your error message and output
